A. REPRODUCIBLE NORMALIZATION PROBLEM

```python
np.random.seed(2112)
X = np.random.randint(10, 101, size=(5, 5))
```

```python
X_normalized = (X - X.mean())/X.std()
```

B. CUBES DIVISIBLE BY 4 PROBLEM

```python
C = np.arange(1,101,1)
```

```python
C = C.reshape(10,10)
```

```python
C = pow(C, 3)
```

```python
div_by_4 = C[C % 4 == 0]
```

```python
np.shape(C)
```

```python
C
```

```python
div_by_4
```

```python
len(div_by_4)
```

```python
np.save("div_by_4.npy", div_by_4)
```

C. ABOVE-MEAN SQUARES PROBLEM

```python
S = np.arange(1, 37, 1)
S.reshape(6, 6)
```

```python
S = pow(S, 2)
```

```python
S_mean = np.mean(S)
```

```python
above_mean = S[S > S_mean]
```

```python
S
```

```python
S_mean
```

```python
above_mean
```

```python
len(above_mean)
```

```python
np.save("above_mean.npy", above_mean)
```
